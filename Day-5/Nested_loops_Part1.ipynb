### Pattern banane ka formula

1. Kitni rows hain? (outer loop)
2. Har row me kitne columns/stars/numbers? (inner loop ki range)
3. Row number i ke saath columns ka kya relation hai? (i, n - i, 2*i - 1)
4. Spaces chahiye? Unke liye alag inner loop.
5. Har row ke baad print() (nayi line).

### Common mistakes

1. Row ke baad print() bhoolna (sab ek line me aa jaata hai).
2. Inner loop ki range me row number i use na karna.
3. Digit loop me original n badal dena aur baad me use karna (pehle copy rakho: temp = n).
4. print(end=" ") aur print() ka confusion.

```python
#🟡 Fix (digit sum hamesha 0 aa raha hai):

n = 123
s = 0
while n > 0:
    s = s+ n % 10
    n = n // 10
print(s)
```

```python
#🟠 Fix (number ka reverse chahiye, 321 ki jagah galat aa raha hai):
n = 123
rev = 0
while n > 0:
    rev = rev*10 + n % 10
    n = n // 10
print(rev)
```

### Section C: Star Patterns (Code Writing)

```python
# 🟢 Square (n x n stars)
for i in range(5):
    for j in range(5):
        print("*",end="")
    print()
```

```python
#🟢 Rectangle (3 rows, 6 columns)

for i in range(3):
    for j in range(6):
        print('*',end="")
    print()
```

```python
'''🟢 Right triangle:
*
**
***
****
*****
'''

for i in range(6):
    for j in range(i):
        print("*",end="")
    print()
```

```python
'''🟡 Inverted triangle (5 stars se 1 star tak).'''

for i in range(5,0,-1):
    for j in range(j):
        print("*",end="")
    print()

```

```python
"""🟡 Right-aligned triangle (spaces + stars):
    *
   **
  ***
 ****
*****
"""

n=5
for i in range(1,n+1):
    print(" "*(n-i) + "*" * i)
```

```python
n=5
for i in range(1, n+1):
    for j in range(n-i):
        print(" ",end="")
    for k in range(i):
        print("*",end="")
    print()
    
```
