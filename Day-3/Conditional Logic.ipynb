```python
Section D: Real-Life Logic
🟡 Leap year: year input lo. Rule: 4 se divisible, lekin 100 se divisible ho to 400 se bhi hona chahiye.
🟡 Grade calculator: marks 90+ A, 75-89 B, 60-74 C, 40-59 D, 40 se kam Fail. (Marks 0-100 ke bahar ho to "Invalid".)
🟡 Electricity bill: 100 units tak ₹5/unit, 101-200 ke liye ₹7/unit (sirf extra units par), 200 ke baad ₹10/unit. Total bill nikalo.
🟡 Salary: basic salary input lo. 20000 se kam ho to bonus 10%, warna 5%. Total salary print karo.
🟡 Age category: 0-12 Child, 13-19 Teen, 20-59 Adult, 60+ Senior.
🟠 Triangle check: teen sides input lo. Valid triangle hai ya nahi, aur valid ho to type (Equilateral, Isosceles, Scalene).
```

```python
# Leap year: year input lo. Rule: 4 se divisible, lekin 100 se divisible ho to 400 se bhi hona chahiye.

year = int(input("Enter year: "))

if year % 400 == 0:
    print("Leap Year")
elif year % 100 == 0:
    print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not a Leap Year")

    
```

```python
# 🟡 Grade calculator: marks 90+ A, 75-89 B, 60-74 C, 40-59 D, 40 se kam Fail. (Marks 0-100 ke bahar ho to "Invalid".)
marks =int(input())
if marks>=100 or marks<=0:
    print("invalid")
elif marks>=90:
    print("A")
elif marks>=75 and marks<=89:
    print("B")
elif marks>=60 and marks<=74:
    print("C")
elif marks>=40 and marks<=59:
    print("D")
elif marks<40:
    print("fail")


    
```

```python
#🟡 Electricity bill: 100 units tak ₹5/unit, 101-200 ke liye ₹7/unit (sirf extra units par), 200 ke baad ₹10/unit. Total bill nikalo.

unit=int(input())
if unit<= 100:
    print("total bill",unit*5)
elif unit>= 101 and unit<=200:
    c=(100*5) + ((unit-100)*7)
    print("total bill",c)
elif unit>=200:
    print("total bill", unit*10)
```

```python
# 🟡 Salary: basic salary input lo. 20000 se kam ho to bonus 10%, warna 5%. Total salary print karo.
sal = int(input())
if sal<=2000:
    print("total salary", sal*0.1+sal)
else:
    print("total salary", sal*0.5+sal)
```

```python
#🟠 Triangle check: teen sides input lo. Valid triangle hai ya nahi, aur valid ho to type (Equilateral, Isosceles, Scalene).

a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
    print("Valid Triangle")

    if a == b == c:
        print("Equilateral")
    elif a == b or b == c or a == c:
        print("Isosceles")
    else:
        print("Scalene")
else:
    print("Invalid Triangle")

'''
Rules:

Equilateral: Teeno sides equal.

Isosceles: Koi do sides equal.

Scalene: Teeno sides different.

Valid triangle: Kisi bhi do sides ka sum, teesri side se greater hona chahiye.'''
```

```python
# 🔴 Shipping Charge: Input: order amount aur city type (metro ya other). 
#Rules: amount 1000+ ho to free shipping. Warna metro ke liye ₹40, other ke liye ₹70. Output: final shipping charge.

amo = int(input())
city_type = input("enter city type--->> metro or other")
if amo==1000:
    print("free shipping")
elif amo>1000 and city_type=="metro":
    print("final shipping charge", amo+40)
elif amo>1000 and city_type=="other":
    print("final shipping charge", amo+70)
```
