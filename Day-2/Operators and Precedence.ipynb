```python
'''1. Difference between / and //
Both are division operators in Python, but they work differently.

/ → Regular division; returns a floating-point result.

// → Floor division; rounds the result down toward negative infinity.'''
```

```python
'''2. == vs is: When can they give different results?
== checks whether two values are equal.

is checks whether two variables refer to the same object in memory.

Example:

python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)

Output:
True
False'''
```

```python
'''3. MCQ: print(3 * "ab" + "c")
Options:

(a) abc

(b) ababab c

(c) abababc

(d) Error

Correct answer: (c) abababc'''
```

```python
'''4. What is short-circuit evaluation in and / or?
Short-circuit evaluation means Python stops evaluating an expression as soon as the result is already determined.


Operator(and)-->> When it finds a falsy value

Operator(or)-->>>When it finds a truthy value'''

```

```python
# 🟠 Day After N Days: Input: today's day number (0 = Monday ... 6 = Sunday) and N. Output: day number after N days. Sample: 5 10 → 1

day_num = int(input("enter day num:"))
N = int(input(" n days"))
ans= (day_num + N )%7
print("Day after N Days:",ans)
```
