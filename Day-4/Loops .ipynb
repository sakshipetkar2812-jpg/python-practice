### Python Day 4: Loops (for, while, range, break, continue, pass)

##1. for loop (jab pata ho kitni baar chalana hai ya kisi sequence par chalana hai)

for i in range(5):
    print(i)        # 0 1 2 3 4


2. range()
range(5) → 0 se 4
range(2, 6) → 2 se 5
range(1, 10, 2) → 1, 3, 5, 7, 9 (step 2)
range(5, 0, -1) → 5, 4, 3, 2, 1
End value include nahi hota.


3. while loop (jab condition True rehne tak chalana ho)
i = 1
while i <= 5:
    print(i)
    i += 1          # update bhoolna mat, warna infinite loop
4. break: loop turant band.
5. continue: is round ko skip karke agle round par. 
6. pass: kuch nahi karta, sirf khali jagah bharta hai.



Common mistakes

while me counter update na karna (infinite loop).
range(1, n) likhna jab n bhi chahiye (sahi: range(1, n + 1)).
Loop ke andar print ka indentation galat.
Loop ke andar variable ko reset kar dena (jaise total = 0 loop ke andar).



Pattern for sum/count/product
total = 0
for i in range(1, 6):
    total += i
print(total)        # 15
for vs while: for = known count / sequence. while = unknown count / condition based.

Real-world use: bill items add karna, ATM retry attempts, data process karna, menu repeat karna.

```python
# 🟢 1 se n tak ka sum nikalo.

n=int(input())
sum=0
for i in range(n+1):
    sum=i+sum
print(sum)
    
```

```python
#🟡 1 se n tak ke numbers ka product (factorial) nikalo.
n=int(input())
fact=1
for i in range(1,n+1):
    fact=fact*i
print(fact)
    
```

```python
# 🟡 1 se 100 tak me kitne numbers 7 se divisible hain? Count print karo.
count=0
for i in range(1,101):
    if i%7==0:
        count=count+1
print(count)
    
```

```python
# 🟡 User se numbers lo jab tak woh 0 na daale (while loop). Sabka sum print karo.

n = int(input())
sum=0
while n!=0:
    sum=sum+n
    n=int(input())
print(sum)
    
```

```python
#🟡 n input lo. Uske saare divisors print karo.
n = int(input())
for i in range(1,n+1):
    if n%i==0:
        print(i)
```

```python
# 🟠 Ek number input lo. Uske digits ki sankhya (count) nikalo bina len() ke (hint: baar-baar // 10).

n=int(input())
count=0

if n==0:
    count=1
else:
    while n>0:
        count=count+1
    n=n//10

print(count)
    
    
```

### Section F: Technical Test Style

```python
# 🔴 Sum of Even and Odd: Input: n.
#  Output: 1 se n tak even numbers ka sum aur odd numbers ka sum, alag-alag. Sample: 10 → 30 25.

n = int(input())
sum_even=0
sum_odd=0
for i in range(1,n+1):
    if i%2==0:
        sum_even=sum_even+i
print(sum_even)
for i in range(1,n+1):
    if i%2!=0:
        sum_odd=sum_odd+i
print(sum_odd)

    
```

```python
#🔴 Count Digits Greater Than 5: Input: ek positive integer. Output: kitne digits 5 se bade hain. Sample: 78234 → 2.

n = int(input())
count=0
while n>0:
    digit=n%10
    if digit >5:
        count+=1
    n=n//10
print(count)

    
```

### Section G: Challenge

```python
#🟠 Number Guessing Game: Program ek secret number (jaise 7) rakhta hai. User guess karta hai.
#  Har galat guess par "Too high" ya "Too low" batao. Sahi hone par kitne attempts me guess kiya, wo print karo.

secret=7
attempts=0

while True:
    guess=int(input("Guess a number : "))
    attempts=attempts+1

    if guess==secret:
        print(f"Right answer {guess}")
        break
    elif guess> secret:
        print("Too High")
    else:
        print("Too Low")
```
