```python
# Is Python compiled or interpreted? Explain briefly.

'''Python is both compiled and interpreted.

Compilation: Python source code (.py) is first compiled into bytecode (.pyc), usually by the Python interpreter.

Interpretation: The Python Virtual Machine (PVM) executes that bytecode.

Example: When you run python hello.py, Python typically compiles the source code into bytecode and then executes it using the PVM.

In short: Python is commonly called an interpreted language, but it uses compilation as part of its execution process'''
```

```python
#  What does "dynamically typed" mean? Give an example.
'''Dynamically typed means that Python determines a variable’s type at runtime, so you don’t need to declare its type explicitly.

Example:

python
x = 10        # x is an integer
x = "Hello"   # now x is a string
x = 3.14      # now x is a float'''
```

```python
# What is the difference between = and ==?

'''The difference between = and == in Python is:

= (Assignment operator): Assigns a value to a variable.

== (Comparison operator): Checks whether two values are equal. It returns True or False.

Example
python
x = 10       # Assign 10 to x

print(x == 10)  # Check if x equals 10
print(x == 5)   # Check if x equals 5'''
```

```python
#  Why does input() return a string? What problem can this cause?

'''In Python, input() returns a string (str) because it treats everything the user types as text, regardless of whether the input looks like a number.

Example
python
age = input("Enter your age: ")
print(type(age))
Output (if the user enters 20):
text
<class 'str'>

What problem can this cause?
You may get unexpected results when performing mathematical operations because the input is a string, not a number.

python
a = input("Enter first number: ")   # User enters 10
b = input("Enter second number: ")  # User enters 20

print(a + b)
Output:

text
1020
Python joins the strings instead of adding the numbers.'''
```
